# J1 — Pitch Types and Called-Pitch Accuracy

This analysis summarizes called-pitch accuracy by pitch type, speed, movement, and zone region. Groups with fewer than 500 called pitches are omitted.

False-strike rate uses called strikes as its denominator. False-ball rate uses called balls. Challenge rate is challenges divided by called pitches in 2026; overturn rate is overturned challenges divided by challenges. Rates include 95% Wilson intervals.

Speed bins are 5 mph wide; movement bins are 5 inches wide. Statcast added the ST and SV pitch labels in 2023, so earlier slider categories may include those pitch types. Comparisons between 2025 and 2026 are descriptive: MLB changed how it measures the zone and pitch location in 2026.

```python
from pathlib import Path
import subprocess
import sys

PROJECT_DIR = Path.cwd()

if not (PROJECT_DIR / "src" / "pitch_type_analysis.py").is_file():
    PROJECT_DIR = PROJECT_DIR.parent

SCRIPT_PATH = PROJECT_DIR / "src" / "pitch_type_analysis.py"

if not SCRIPT_PATH.is_file():
    raise FileNotFoundError(f"Could not find {SCRIPT_PATH}")

subprocess.run(
    [sys.executable, str(SCRIPT_PATH)],
    cwd=PROJECT_DIR,
    check=True,
)
```

```python
import pandas as pd
from IPython.display import Image, display

TABLES_DIR = PROJECT_DIR / "reports" / "tables"
FIGURES_DIR = PROJECT_DIR / "reports" / "figures"

reports = [
    ("pitch_types_by_type.csv", "Pitch type"),
    ("pitch_types_by_speed.csv", "Speed band"),
    ("pitch_types_by_horizontal_movement.csv", "Horizontal movement"),
    ("pitch_types_by_vertical_movement.csv", "Vertical movement"),
    ("pitch_types_by_zone_region.csv", "Zone region"),
]

for filename, title in reports:
    print(title)
    table = pd.read_csv(TABLES_DIR / filename)
    display(
        table.loc[table["season"].eq(2026)]
        .sort_values("error_rate", ascending=False)
    )

display(Image(filename=str(
    FIGURES_DIR / "pitch_types_error_rate_2026.png"
)))
```
