# J3 — Within-Umpire Accuracy and Called Zone

The analysis compares each plate umpire’s accuracy in 2026 with their pooled accuracy from 2021–2025. It includes umpires with at least 1,500 called pitches in both periods. Accuracy rates have 95% Wilson intervals; the 2026-minus-prior difference has a 95% interval from 2,000 game-level bootstrap samples.

The called-zone estimate groups pitches into 1-inch cells using absolute plate location. It traces the largest contour where at least 50% of calls are strikes. Cells with fewer than 100 called pitches are excluded.

The 2026 ABS reference is 17 inches wide, with vertical limits summarized by the median recorded zone bottom and top. MLB measured plate coordinates at the front through 2025 and at the midpoint in 2026, so comparisons across seasons are descriptive.

```python
from pathlib import Path
import subprocess
import sys

PROJECT_DIR = Path.cwd()

if not (PROJECT_DIR / "src" / "j3_called_zone.py").is_file():
    PROJECT_DIR = PROJECT_DIR.parent

SCRIPT_PATH = PROJECT_DIR / "src" / "j3_called_zone.py"

if not SCRIPT_PATH.is_file():
    raise FileNotFoundError(f"Could not find {SCRIPT_PATH}")

result = subprocess.run(
    [sys.executable, str(SCRIPT_PATH)],
    cwd=PROJECT_DIR,
    check=True,
    capture_output=True,
    text=True,
)

print(result.stdout)
if result.stderr:
    print(result.stderr)
```

```python
import pandas as pd
from IPython.display import Image, Markdown, display

TABLES_DIR = PROJECT_DIR / "reports" / "tables"
FIGURES_DIR = PROJECT_DIR / "reports" / "figures"
REPORTS_DIR = PROJECT_DIR / "reports"

umpires = pd.read_csv(
    TABLES_DIR / "j3_within_umpire_accuracy.csv"
)
umpire_columns = [
    "official_name",
    "prior_called_pitches",
    "prior_accuracy",
    "prior_ci_low",
    "prior_ci_high",
    "season_2026_called_pitches",
    "season_2026_accuracy",
    "season_2026_ci_low",
    "season_2026_ci_high",
    "accuracy_change",
    "change_ci_low",
    "change_ci_high",
]

print("Largest accuracy increases")
display(
    umpires.sort_values("accuracy_change", ascending=False)
    .head(10)[umpire_columns]
)

print("Largest accuracy decreases")
display(
    umpires.sort_values("accuracy_change")
    .head(10)[umpire_columns]
)

print("Called-zone edges by season")
display(
    pd.read_csv(TABLES_DIR / "j3_called_zone_edges.csv")
)

display(
    Image(
        filename=str(
            FIGURES_DIR / "j3_umpire_accuracy_change.png"
        ),
        width=700,
    )
)

display(
    Markdown(
        (REPORTS_DIR / "j3_called_zone_notes.md").read_text(
            encoding="utf-8"
        )
    )
)
```
