# J5: Umpire-Season Correlations

This analysis explores how regular-season umpire accuracy relates to workload, experience, crew history, and 2026 challenge outcomes. Correlations describe associations; they do not establish causes. The same umpire can appear in multiple seasons, and MLB changed how it measures the strike zone in 2026.

```python
from pathlib import Path
import subprocess
import sys

import pandas as pd
from IPython.display import Markdown, display

project_dir = Path.cwd()
while not (project_dir / "src" / "j5_correlations.py").exists():
    if project_dir.parent == project_dir:
        raise FileNotFoundError("Could not locate the project folder.")
    project_dir = project_dir.parent

subprocess.run(
    [sys.executable, str(project_dir / "src" / "j5_correlations.py")],
    cwd=project_dir,
    check=True,
)

panel = pd.read_csv(
    project_dir / "reports" / "tables" / "j5_umpire_season.csv"
)
correlations = pd.read_csv(
    project_dir / "reports" / "tables" / "j5_correlation_matrix.csv",
    index_col=0,
)
sample_sizes = pd.read_csv(
    project_dir / "reports" / "tables" / "j5_correlation_sample_sizes.csv",
    index_col=0,
)
notes = (
    project_dir / "reports" / "j5_correlation_notes.md"
).read_text(encoding="utf-8")

display(Markdown("## Umpire-season data preview"))
display(
    panel[
        [
            "season",
            "official_name",
            "accuracy",
            "called_pitches",
            "games_worked",
            "distinct_crew_count",
            "age_at_season_start",
            "years_since_mlb_debut",
            "challenge_rate_2026",
            "overturn_rate_2026",
        ]
    ].head(10)
)

display(Markdown("## Pearson correlation matrix"))
display(correlations)

display(Markdown("## Pairwise sample sizes"))
display(sample_sizes)

display(Markdown(notes))
```

## Key observations

- Age and years since MLB debut were strongly correlated (r = 0.96, n = 460), so their separate relationships with accuracy should be interpreted cautiously.
- Across the pooled umpire-seasons, accuracy was negatively associated with age (r = -0.49, n = 461) and MLB experience (r = -0.50, n = 508). These measures also move with calendar time.
- For 2026, accuracy was negatively associated with challenge rate (r = -0.57, n = 89) and overturn rate (r = -0.72, n = 89). Overturn rate is based only on challenged pitches, so this is a descriptive association.
- Called pitches and plate games were strongly correlated (r = 0.99, n = 524). The pairwise sample sizes vary because some ages and MLB debut dates are missing, and challenge measures are available only for 2026.
- These correlations do not establish causes. The 2025-to-2026 accuracy comparison also mixes umpire behavior with MLB’s change in how it measures the strike zone.
